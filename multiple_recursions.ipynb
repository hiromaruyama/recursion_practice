Problem 1: Write a recursive function count_coins that takes a positive integer total and a non-negative integer k and returns the number of ways to make change for total using exactly k coins.


```python
def next_smaller_coin(coin):
  if coin == 25:
    return 10
  elif coin == 10:
    return 5
  elif coin == 5:
    return 1

def inner(total, coin, k):
  if total ==0:
    if k == 0:
      return 1
    return 0
    if coin == None:
        return 0
    if total<0:
        return 0

    use = inner(total-coin, coin, k-1)
    skipped = inner(total, next_smaller_coin(coin), k)
    return use + skipped
  return inner(total, 25, k)

print(count_coins(15, 3))
print(count_coins(15, 4))
print(count_coins(30, 6))
```

Problem 2: Write a recursive function count_coins that takes a positive integer total and a non-negative integer k and returns the number of ways to make change for total using at least k non-quarter coins.

```python
def next_smaller_coin(coin):
    """Returns the next smaller coin in order."""
    if coin == 25:
        return 10
    elif coin == 10:
        return 5
    elif coin == 5:
        return 1


def count_coins(total, k):
    """Return the number of ways to make change for total using at least k
    non-quarter coins (1-cent, 5-cent, and 10-cent)."""
    # YOUR CODE HERE
    def inner(total, coin, k):
        if total ==0:
            if k <=0:
              return 1
            return 0
        if total <0:
          return 0
        if coin== None:
          return 0
        if coin != 25:
          use = inner(total-coin, coin, k-1)
        else:
          use = inner(total-coin, coin, k)
        skipped = inner(total, next_smaller_coin(coin), k)
        return use + skipped
    return inner(total, 25, k)
```

Problem 3: Write a recursive function count_coins that takes a positive integer total and returns the number of ways to make change for total using an odd number of coins.

```python
def next_smaller_coin(coin):
    """Returns the next smaller coin in order."""
    if coin == 25:
        return 10
    elif coin == 10:
        return 5
    elif coin == 5:
        return 1

def count_coins(total):
    """Return the number of ways to make change for total using an odd number of
    coins."""
    # YOUR CODE HERE
    def inner(total, coin, k):
      if total == 0:
        if k % 2 != 0:
          return 1
        else:
          return 0
      if coin == None:
        return 0
      if total<0:
        return 0
      use = inner(total-coin, coin, k+1)
      skipped = inner(total, next_smaller_coin(coin), k)
      return use + skipped

    return inner(total, 25, 0)

print(count_coins(10)) #1
print(count_coins(15)) #4
```

Question 4: Write a recursive function count_coins that takes a positive integer total and a non-negative integer k and returns the number of ways to make change for total using at most k of each type of coin.

```python
def next_smaller_coin(coin):
    """Returns the next smaller coin in order."""
    if coin == 25:
        return 10
    elif coin == 10:
        return 5
    elif coin == 5:
        return 1


def count_coins(total, k):
    """Return the number of ways to make change for total using at most k of each
    type of coin."""
    # YOUR CODE HERE
    def inner(total, coin, count):
      if count>k:
        return 0
      if total == 0:
        return 1
      if total<0:
        return 0
      if coin == None:
        return 0

      use = inner(total-coin, coin, count+1)

      skipped = inner(total, next_smaller_coin(coin), 0)
      return use + skipped

    return inner(total, 25, 0)



print(count_coins(15, 5))
print(count_coins(15, 15))
```

7. Write a recursive function count_coins that takes a positive integer total and a non-negative integer k and returns the number of ways to make change for total using at least k small coins.

```python
def next_smaller_coin(coin):
    """Returns the next smaller coin in order."""
    if coin == 25:
        return 10
    elif coin == 10:
        return 5
    elif coin == 5:
        return 1


def count_coins(total, k):
    """Return the number of ways to make change for total using at least k small
    coins (1-cent and 5-cent)."""
    # YOUR CODE HERE
    def inner(total, coin, counting):
        if total == 0:
            if counting>=k:
                return 1
            else:
                return 0

        if coin == None:
            return 0
        if total<0:
            return 0

        if coin == 1 or coin== 5:
            use = inner(total-coin, coin, counting+1)
        else:
            use = inner(total-coin, coin, counting)

        skipped = inner(total, next_smaller_coin(coin), counting)
        return use + skipped
    return inner(total, 25, 0)
```
